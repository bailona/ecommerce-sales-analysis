```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("🚀 Data Science Project Started!")
```

```python
data = {
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headset"],
    "sales": [120, 350, 200, 80, 150],
    "price": [4500, 120, 250, 1800, 350]
}

df = pd.DataFrame(data)

df
```

```python
df["revenue"] = df["sales"] * df["price"]

df
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["price"], df["sales"])

for i, product in enumerate(df["product"]):
    plt.annotate(
        product,
        (df["price"].iloc[i], df["sales"].iloc[i]),
        xytext=(8, 5),
        textcoords="offset points"
    )

plt.title("Price vs Units Sold")
plt.xlabel("Price (R$)")
plt.ylabel("Units Sold")

plt.show()
```

```python
import numpy as np

x = df["price"]
y = df["sales"]

coefficients = np.polyfit(x, y, 1)

slope = coefficients[0]
intercept = coefficients[1]

print(f"Slope: {slope:.4f}")
print(f"Intercept: {intercept:.2f}")
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["price"], df["sales"])

for i, product in enumerate(df["product"]):
    plt.annotate(
        product,
        (df["price"].iloc[i], df["sales"].iloc[i]),
        xytext=(8, 5),
        textcoords="offset points"
    )

x_line = np.linspace(df["price"].min(), df["price"].max(), 100)
y_line = slope * x_line + intercept

plt.plot(x_line, y_line)

plt.title("Price vs Units Sold")
plt.xlabel("Price (R$)")
plt.ylabel("Units Sold")

plt.show()
```

# Conclusion

## Key Findings

- The Laptop generated the highest revenue.
- The Mouse had the highest number of units sold.
- Revenue was highly concentrated in the Laptop product.
- The analysis showed a negative relationship between price and units sold in this small sample.
- The linear regression is illustrative and should not be used as a business prediction model because the dataset contains only five products.

```python
total_revenue = df["revenue"].sum()
best_revenue_product = df.loc[df["revenue"].idxmax(), "product"]
best_selling_product = df.loc[df["sales"].idxmax(), "product"]
average_price = df["price"].mean()

print(f"Total Revenue: R$ {total_revenue:,.2f}")
print(f"Top Revenue Product: {best_revenue_product}")
print(f"Best-Selling Product: {best_selling_product}")
print(f"Average Product Price: R$ {average_price:,.2f}")
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    df["revenue"],
    labels=df["product"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Revenue Distribution by Product")
plt.show()
```

# Executive Summary

## Business Insights

This analysis explored product sales, prices, and revenue.

### Key Findings

- Total revenue was R$ 828,500.
- Laptop generated the highest revenue, with approximately 65% of total revenue.
- Mouse was the best-selling product by units, with 350 units sold.
- The average product price was R$ 1,404.
- The analysis showed a negative relationship between product price and units sold in this small dataset.
- The dataset contains only five products, so the results should be considered exploratory rather than statistically conclusive.

## Main Business Insight

The product with the highest sales volume was not the product generating the highest revenue.

This demonstrates why analyzing both sales volume and revenue is important when evaluating product performance.
